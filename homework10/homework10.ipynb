```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
from astropy import units as u 
from astropy import constants as const 

plt.rcParams["figure.figsize"] = (6, 4)
plt.rcParams["figure.dpi"] = 120
```

```python
csv_path = "solar_system.csv"
df = pd.read_csv(csv_path)

print("Original DataFrame shape:"   , df.shape) 
display(df.head())
```

```python
# Cell 4: transpose the DataFrame so rows become planets instead of attributes

# Save shape BEFORE
shape_before = df.shape
print("Shape BEFORE transposing:", shape_before)

# 1) Set "Attribute" as index so each attribute becomes the row index
df = df.set_index("Attribute")
print("Shape after set_index('Attribute'):", df.shape)

# 2) Transpose so attributes become columns and planets become rows
df = df.T
print("Shape after transpose .T:", df.shape)

# 3) Name the index "Planet"
df.index.name = "Planet"
print("Index name:", df.index.name)

# 4) Reset index so "Planet" becomes its own column
df.reset_index(inplace=True)
print("Shape after reset_index:", df.shape)

# 5) Remove columns.name if set
df.columns.name = None

shape_after = df.shape
print("Final shape AFTER transposing:", shape_after)

display(df.head())
```


**(c)** What each line of code does:
- `df = df.set_index("Attribute")`  
  Uses the `"Attribute"` column as the index. The labels in that column become row labels instead of a normal column.
- `df = df.T`  
  Transposes the DataFrame: rows become columns and columns become rows.
- `df.index.name = "Planet"`  
  Assigns the name `"Planet"` to the index so it’s clear that each row corresponds to a planet.
- `df.reset_index(inplace=True)`  
  Turns the index (`Planet`) back into a normal column called `"Planet"`, and creates a new default integer index.
- `df.columns.name = None`  
  Clears the name of the columns axis; purely cosmetic.

**(d)**   the original table’s structure is not exactly 11 planets by 20 attributes
- There might be extra rows that aren’t planets 
- Some attributes might not be planetary parameters, or there might be extra columns.  
So after transposing, the number of rows and columns is determined by whatever was actually in the original CSV, not just planets vs attributes

**(f)** 
- “Columns with units” are those whose names contain things like `(km)`, `(m/s^2)`, `(10^6 km)`, `(kg)`, `(days)`, `(deg)`
- “Columns without units” are dimensionless (like `Planet`, `Number of Moons`, `Ring System?`, `Global Magnetic Field?`, `Orbital eccentricity`).




```python

print("Column names:\n")
print(df.columns.tolist())

cols_with_units = []
cols_without_units = []

for col in df.columns:
    if any(unit in col for unit in ["(", ")", "km", "kg", "m/s", "days", "deg"]):
        cols_with_units.append(col)
    else:
        cols_without_units.append(col)

print("\nColumns w/ units in their names:")
for col in cols_with_units:
    print("  -", col)

print("\nColumns that don't have units in their names:")
for col in cols_without_units:
    print("  -", col)

print("\nFull transposed DataFrame:")
display(df)
```

```python

print("Column dtypes from pandas:\n")
print(df.dtypes)

print("\nExample value types inside each column:\n")
for col in df.columns:
    print(f"Column: {col}")
    print(df[col].apply(type).value_counts(), "\n")
```

## 2.1.2 

**(a)** 
From the `df[col].apply(type)` output, most entries are probably `<class 'str'>` (strings).  
After we convert, they will be numeric types like `float` or `int` for most columns.

**(b)** 
- You cannot do math directly on strings; `"10" + "5"` becomes `"105"`, not `15`.  


```python

non_numeric_cols = ["Planet", "Ring System?", "Global Magnetic Field?"]

for col in df.columns:
    if col not in non_numeric_cols:
        df[col] = pd.to_numeric(df[col], errors="coerce")

print("New dtypes after conversion:\n")
print(df.dtypes)

object_columns = df.select_dtypes(include="object").columns.tolist()
print("\nColumns still of dtype 'object':", object_columns)
```

```python

def attach_units(df, old_col, unit, new_col):
    numeric_values = pd.to_numeric(df[old_col], errors="coerce")
    
    quantities = numeric_values.values * unit
    
    df[new_col] = quantities
 
    df.drop(columns=[old_col], inplace=True)

    print(f"Attached units {unit} to column '{old_col}' → '{new_col}'")
    return df
```

```python
from astropy import units as u

def attach_units(df, old_col, unit, new_col):
    
    new_values = []

    for val in df[old_col]:
        new_values.append(val * unit)

    df[old_col] = new_values

    df.rename(columns={old_col: new_col}, inplace=True)

```

```python
unit    = 1e24 * u.kg
old_col = "Mass (10^24kg)"   
new_col = "Mass (kg)"

attach_units(df, old_col, unit, new_col)

df[["Planet", new_col]].head()
```

```python

peri_col = "Perihelion (10^6 km)"
aphe_col = "Aphelion (10^6 km)"

df["Perihelion (km)"] = df[peri_col].values * (1e6 * u.km)
df["Aphelion (km)"]   = df[aphe_col].values * (1e6 * u.km)

semi_major = (df["Perihelion (km)"] + df["Aphelion (km)"]) / 2

aphe_loc = df.columns.get_loc(aphe_col)
new_col_name = "Semi-major axis (km)"

df.insert(aphe_loc + 1, new_col_name, semi_major)

print("Column names after inserting semi-major axis:")
print(df.columns.tolist())

print("\nSemi-major axis column:")
print(df[["Planet", new_col_name]].head())
```

```python
orbital_period_col = "Orbital Period (days)"

period_days = df[orbital_period_col].values * u.day

period_years = period_days.to(u.year)

new_period_col = "Orbital Period (yr)"

df[new_period_col] = period_years
print("column names after converting orbital period:")
print(df.columns.tolist())

print("\nFull orbital period (yr) column:")
print(df[["Planet", new_period_col]])

favorite_planet ="Jupiter"
row = df[df["Planet"] == favorite_planet]

orbital_period = row[new_period_col].values[0]
print(f"\n{favorite_planet}'s orbital period is {orbital_period:.4f} yr")
```
