## Cell 1 — Imports

```
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

## Cell 2 — Load the data file with NumPy
- Adjust the relative path if your folders differ.
- `skiprows=1` ignores the header line.

If your class repo path is different, change it here.

```
data_path = "../../course_assignments/homework7/star_data.txt"  # adjust if needed
data = np.loadtxt(data_path, skiprows=1)
print("Data loaded. Shape:", data.shape)
```

## Cell 3 — Split columns into variables
- Column 0: magnitude (bigger number = dimmer)
- Column 1: temperature (Kelvin)

```
magnitude = data[:, 0]
temperature = data[:, 1]
print("Magnitude min/max:", magnitude.min(), magnitude.max())
print("Temperature min/max:", temperature.min(), temperature.max())
```

## Cell 4 — Find indices of brightest/dimmest and hottest/coldest
- Remember: **smaller** magnitude → **brighter** star (mags are "backwards").

```
idx_brightest = np.argmin(magnitude)  # smallest magnitude
idx_dimmest   = np.argmax(magnitude)  # largest magnitude
idx_hottest   = np.argmax(temperature)
idx_coldest   = np.argmin(temperature)

print("Brightest star: index", idx_brightest, "mag=", magnitude[idx_brightest])
print("Dimmest star:   index", idx_dimmest,   "mag=", magnitude[idx_dimmest])
print("Hottest star:   index", idx_hottest,   "T=", temperature[idx_hottest])
print("Coldest star:   index", idx_coldest,   "T=", temperature[idx_coldest])
```

## Cell 5 — Make log versions (natural log for simplicity)

```
log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)
```

## Cell 6 — Scatter plot the HR diagram
- x-axis: log(Temperature), inverted (hotter on the left)
- y-axis: log(Magnitude), inverted (brighter at the top)
- Then highlight special stars with different colors/markers

```
plt.figure(figsize=(8, 6))

# Base scatter: color by temperature (optional colormap bonus)
sc = plt.scatter(log_temperature, log_magnitude, s=10, alpha=0.7, c=temperature)
plt.colorbar(sc, label='Temperature (K)')

# Highlight the four stars
plt.scatter(log_temperature[idx_brightest], log_magnitude[idx_brightest], 
            marker='*', s=300, color='black', label='Brightest (min mag)')
plt.scatter(log_temperature[idx_dimmest], log_magnitude[idx_dimmest], 
            marker='*', s=300, color='purple', label='Dimmest (max mag)')
plt.scatter(log_temperature[idx_hottest], log_magnitude[idx_hottest], 
            marker='*', s=300, color='green', label='Hottest (max T)')
plt.scatter(log_temperature[idx_coldest], log_magnitude[idx_coldest], 
            marker='*', s=300, color='orange', label='Coldest (min T)')

# Invert both axes (astronomy convention)
ax = plt.gca()
ax.invert_xaxis()
ax.invert_yaxis()

plt.title("Hertzsprung–Russell Diagram")
plt.xlabel("Temperature (log K)")
plt.ylabel("Magnitude (log)")
plt.legend(loc='best')
plt.tight_layout()
plt.show()
```
